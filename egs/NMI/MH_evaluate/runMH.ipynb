```python
import os
import shutil
import time

# 1. Define where the files are being saved right now
local_source_dir = "/content/NeuralMAG/NeuralMAG/egs/NMI/MH_evaluate/figs16/shape_True/"

# 2. Define your permanent target folder inside your mounted Google Drive
drive_target_dir = "/content/drive/MyDrive/NeuralMAG_Backups/figs16/shape_True/"
os.makedirs(drive_target_dir, exist_ok=True)

print("Starting background backup loop... Press 'Stop' in VS Code to end it manually.")

# 3. Continuous sync loop
try:
    while True:
        if os.path.exists(local_source_dir):
            files = os.listdir(local_source_dir)
            copied_count = 0
            
            for file_name in files:
                source_file = os.path.join(local_source_dir, file_name)
                target_file = os.path.join(drive_target_dir, file_name)
                
                # Copy the file if it doesn't exist in Drive yet, or if it was recently updated
                if not os.path.exists(target_file) or os.path.getmtime(source_file) > os.path.getmtime(target_file):
                    shutil.copy2(source_file, target_file)
                    copied_count += 1
            
            if copied_count > 0:
                print(f"Successfully backed up {copied_count} new/updated files to Google Drive.")
        else:
            print("Waiting for local directory to be populated by the simulation...")
            
        # Check for updates every 60 seconds
        time.sleep(60)

except KeyboardInterrupt:
    print("\nBackup loop stopped cleanly by user.")
```
